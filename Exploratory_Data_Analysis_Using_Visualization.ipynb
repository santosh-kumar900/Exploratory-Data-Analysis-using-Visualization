**Load Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Import Titanic Data set**

```python
df = pd.read_csv('/content/Titanic.csv')
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.tail()
```

**Handle missing**

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Cabin'] = df['Cabin'].fillna('Unknown')
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.info()
```

### Q1. Create a countplot to show the number of passengers by gender.

```python
gender_counts =df['Sex'].value_counts()
gender_counts
```

```python
from ipywidgets.widgets.widget_int import Color
gender_counts.plot(kind='bar', color=['blue','pink'])
plt.title('Gender Distribution')
plt.xlabel('Sex')
plt.ylabel('Count')
```

### Q2. Create a bar chart to show the number of passengers in each passenger class using different colors.

```python
p_class_count = df['Pclass'].value_counts()
p_class_count
```

```python
p_class_count.plot(kind='bar', color=['red','green','orange'])
plt.title('Passenger class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
```

### Q3. Create a histogram to show the distribution of passenger age

```python
sns.histplot(x='Age', bins=30, data=df, kde=True, color='skyblue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
```

### Q4. Create a boxplot to detect outliers in the fare column.

```python
plt.boxplot(df['Fare'], vert=False, patch_artist=True)
plt.title('Boxplot of fare')
plt.xlabel('Fare Value')
```

### Q5. Create a scatterplot to

**a. show the relationship between age and fare.**

**b. analyze the relationship between family size (SibSp + Parch) and fare.**

```python
# a. show the relationship between age and fare.
plt.scatter(df['Age'], df['Fare'], alpha=0.7, edgecolors='white')
plt.title('Relationship between Age and Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
```

```python
# b. analyze the relationship between family size (SibSp + Parch) and fare.
df['Family_Size'] = df['SibSp'] + df['Parch']

plt.scatter(df['Family_Size'],df['Fare'], alpha=0.7, edgecolors='white', color='green')
```

### Q6. Create a countplot to show survival count by gender.

```python
sns.countplot(data=df, x='Sex', hue='Survived', palette='Set1')
plt.title('Survived Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Passenger Count')
```

### Q7. Create a bar chart to show survival by passenger class.

```python
sns.countplot(data=df, x='Pclass', hue='Survived', palette='Set1')
plt.title('Survived Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Passenger Count')
```

### Q8. Create a heatmap to show correlation between numerical variables.

```python
numerical_df = df.select_dtypes(include=['number'])
corr_matrix = numerical_df.corr()
plt.figure(figsize=(14,10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt ='.2f', linewidths=0.5 )
plt.title('Correlation Heatmap of Numerical Variables', fontsize=16)
```

### Q9. Identify which passenger class paid the highest fare on average.

```python
average_fare = df.groupby('Pclass')['Fare'].mean()
```

```python
print('Average Fare by Passenger Class')
print(average_fare)
```

```python
highest_class = average_fare.idxmax()
highest_fare = average_fare.max()
print(f"\nPassenger Class {highest_class} paid the highest fare on average: {highest_fare:.2f}")
```
